## Visualzation (Histogram)

```r
ggplot(df, aes(x = gnipc)) +                # Refer to data in "df", define X-Axis by variable "grp"
  geom_histogram(bins = 10,               # Number of bars/classes. 
                 fill = "steelblue",      # Bar fill color 
                 color = "white",         # Bar border color
                 alpha = 0.8) +           # Transparency (0-1)
  labs(title = "Distribution of Gross Regional Product per capita across Chinese Provinces 2024",# Chart title
       subtitle = "In 10,000 CNY",        # Chart subtitle
       x = "Gross Regional Product",                   # X-Axix label 
       y = "Count",                       # Y-Axix label  
       caption = "Data Source: National Bureau of Statistics of China") +  # Chart caption
  theme_minimal()                         # Chart general layout
```

## Visualization of Group Statistics by using Boxplots

```r
ggplot(df, aes(x = region4, y = gnipc)) +
  geom_boxplot(alpha = 0.7,                    # Transparency 
               fill = "steelblue") +           # Fill color
  labs(title = "Comparison of Gross Regional Product per capita across Regions in China 2024",
       x = "Region",
       y = "GRP Value (10,000 CNY)",
       caption = "Data Source: National Bureau of Statistics of China") +
  theme_minimal() +
  theme(legend.position = "none")              # Remove legend (redundant here)
```

## Visualization of Group Statistics by using Means and Standard Deviations



```r

# We have to calculate the parameters first and save the group values as new variables
# NOTE: You don't have to fully understand the coding, no worries. :)
tmean = by(gnipc, region4, mean) # compute group means
tsd = by(gnipc, region4, sd) # compute standard deviations of groups
# Merge both parameter sets into a new data set
df_means = data.frame(tmean, tsd) # create data set 
df_means$region = rownames(df_means) # add region labels

# Create the chart
ggplot(df_means, aes(x=region, y=tmean)) + 
  geom_errorbar(aes(ymin = tmean-tsd, ymax = tmean+tsd), width=0.01, color="steelblue") +
  geom_point(color="darkblue", size=3) + 
  labs(title = "Comparison of Gross Regional Product across Regions in China",
       x = "Region",
       y = "GRP Value (10,000 CNY)",
       caption = "Data Source: National Bureau of Statistics of China") +
  theme_minimal()                         # A clean, popular theme

```

## Visualization of the association between two metric variables

```r
ggplot(df, aes(x = gnipc, y = hdi)) +        # Define axes
  geom_point(alpha = 0.7,                    # Transparency of points
             color = "steelblue") +          # Color of points
  labs(title = "Association of GRP in Secondary and Tertiary Sector",
       subtitle = "10,000 CNY",
       x = "GRP II. Sector (Industry)",
       y = "GRP III. Sector (Services)",
       caption = "Data Source: National Bureau of Statistics of China") +
  theme_minimal() +
  theme(legend.position = "none")           # Remove legend
```

## **Additional techniques**

## Regression analysis

```r
lm(hdi ~ gnipc)
# on average, we observe additional 55,920 CNY 
# gain in services productivity for each increase 
# of primary sector poductivity by 10,000 CNY  
```

```r
lm(hdi ~ gnipc, data=df[!(grp3 > 38000 & grp2 < 20000),])
# There is no big change in output if outliers are excuded
```

## Visualization of regression line

```r

ggplot(df, aes(x = gnipc, y = hdi)) +        
  geom_point(alpha = 0.7,                    
             color = "steelblue") +          
  geom_smooth(method = "lm", formula = y ~ x, se = F, color = "darkblue") +  # regression line added
  labs(title = "Association of GRP in Secondary and Tertiary Sector",
       subtitle = "10,000 CNY",
       x = "GRP II. Sector (Industry)",
       y = "GRP III. Sector (Services)",
       caption = "Data Source: National Bureau of Statistics of China") +
  theme_minimal() +
  theme(legend.position = "none")           # Remove legend
```
